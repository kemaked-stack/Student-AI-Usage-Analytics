```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# 1. Load data and clean up string representations of missing values
df = pd.read_csv('students_ai_usage.csv')
df['uses_ai'] = df['uses_ai'].str.strip()

# 2. Calculate the core business metric: Grade Improvement
df['grade_improvement'] = df['grades_after_ai'] - df['grades_before_ai']

print("--- DESCRIPTIVE STATISTICS ---")
print(df.groupby('uses_ai')['grade_improvement'].describe())

# 3. Hypothesis Testing (Independent Samples t-test)
# Testing if the grade improvement of AI users is statistically different from non-users
ai_users = df[df['uses_ai'] == 'Yes']['grade_improvement']
non_users = df[df['uses_ai'] == 'No']['grade_improvement']

t_stat, p_val = stats.ttest_ind(ai_users, non_users, equal_var=False)
print("\n--- HYPOTHESIS TEST RESULTS ---")
print(f"t-statistic: {t_stat:.4f}")
print(f"p-value: {p_val:.4e} (If less than 0.05, the result is highly significant)")

# 4. Visualization 1: AI Usage vs Grade Improvement
plt.figure(figsize=(6, 4))
sns.barplot(x='uses_ai', y='grade_improvement', data=df, errorbar=None, palette='Blues_d')
plt.title('Average Grade Improvement: AI Users vs Non-Users', fontsize=12, fontweight='bold')
plt.xlabel('Uses Generative AI Tools', fontsize=10)
plt.ylabel('Average Grade Improvement (Points)', fontsize=10)
plt.tight_layout()
plt.savefig('ai_vs_grades.png', dpi=300)
plt.show()

# 5. Visualization 2: Purpose of AI vs Grade Improvement (For AI Users Only)
ai_only_df = df[df['uses_ai'] == 'Yes']
plt.figure(figsize=(7, 4))
sns.barplot(x='purpose_of_ai', y='grade_improvement', data=ai_only_df, errorbar=None, palette='YlGnBu_r')
plt.title('Grade Improvement by Primary Purpose of AI Use', fontsize=12, fontweight='bold')
plt.xlabel('Primary Purpose', fontsize=10)
plt.ylabel('Average Grade Improvement (Points)', fontsize=10)
plt.tight_layout()
plt.savefig('purpose_vs_grades.png', dpi=300)
plt.show()

# Run the t-test to check if the grade difference is a random fluke
t_stat, p_val = stats.ttest_ind(ai_users, non_users, equal_var=False)
```
